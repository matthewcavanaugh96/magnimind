# Aggregates and new dataframes

```python
recom_df_with_aggs = recom_df.copy()

# unique products per transaction
recom_df_with_aggs['unique_products_per_transac'] = (recom_df.groupby('Transaction_ID')['Product_ID'].transform('nunique'))

# total items per transaction
recom_df_with_aggs['total_items_per_transac'] = (recom_df.groupby('Transaction_ID')['Product_ID'].transform('count'))

# total spent per transaction
recom_df_with_aggs['total_spent_per_transac'] = (recom_df.groupby('Transaction_ID')['Price'].transform('sum'))

# =====================
# =====================

# unique products per customer
recom_df_with_aggs['unique_products_per_customer'] = (recom_df.groupby('Customer_ID')['Product_ID'].transform('nunique'))

# total items per customer
recom_df_with_aggs['total_items_per_customer'] = (recom_df.groupby('Customer_ID')['Product_ID'].transform('count'))

# total spent per customer
recom_df_with_aggs['total_spent_per_customer'] = (recom_df.groupby('Customer_ID')['Price'].transform('sum'))

# total transactions per customer
recom_df_with_aggs['transac_per_customer'] = (recom_df.groupby('Customer_ID')['Transaction_ID'].transform('nunique'))

# =====================
# =====================

# unique customers per product
recom_df_with_aggs['unique_customers_per_product'] = (recom_df.groupby('Product_ID')['Customer_ID'].transform('nunique'))

# total quantity sold per product
recom_df_with_aggs['total_sold_per_product'] = (recom_df.groupby('Product_ID')['Amount'].transform('sum'))

# total revenue per product
recom_df_with_aggs['total_revenue_per_product'] = (
    recom_df
    .assign(revenue=recom_df['Price'] * recom_df['Amount'])
    .groupby('Product_ID')['revenue']
    .transform('sum')
)

# median price per product
recom_df_with_aggs['median_price_per_product'] = (
    recom_df_with_aggs
        .groupby('Product_ID')['Price']
        .transform('median')
)

# min price per product
recom_df_with_aggs['max_price_per_product'] = (
    recom_df_with_aggs
        .groupby('Product_ID')['Price']
        .transform('max')
)

# max price per product
recom_df_with_aggs['min_price_per_product'] = (
    recom_df_with_aggs
        .groupby('Product_ID')['Price']
        .transform('min')
)

# price vs median
recom_df_with_aggs['price_vs_median'] = recom_df_with_aggs['Price'] - recom_df_with_aggs['median_price_per_product']

# median spend per transaction per customer
recom_df_with_aggs['median_spend_per_transaction_per_customer'] = (
    recom_df_with_aggs
        .groupby('Customer_ID')['total_spent_per_transac']
        .transform('median')
)

recom_df_with_aggs['max_spend_per_transaction_per_customer'] = (
    recom_df_with_aggs
        .groupby('Customer_ID')['total_spent_per_transac']
        .transform('max')
)

recom_df_with_aggs['min_spend_per_transaction_per_customer'] = (
    recom_df_with_aggs
        .groupby('Customer_ID')['total_spent_per_transac']
        .transform('min')
)



# All products bought by customer
customer_products = (
    recom_df_with_aggs
        .groupby('Customer_ID')['Product_ID']
        .agg(lambda x: sorted(x.dropna().unique()))
)
recom_df_with_aggs['products_bought_by_customer'] = (
    recom_df_with_aggs['Customer_ID'].map(customer_products)
)


# For each product, a count of repeat customers
product_customer_counts = (
    recom_df_with_aggs
        .groupby(['Product_ID', 'Customer_ID'])
        .size()
)
repeat_customers_per_product = (
    product_customer_counts
        .gt(1)                      # True if repeat buyer
        .groupby('Product_ID')
        .sum()                      # count of True values
)
recom_df_with_aggs['repeat_customers_per_product'] = (
    recom_df_with_aggs['Product_ID'].map(repeat_customers_per_product)
)

# Pct repeat customers per product
recom_df_with_aggs['pct_repeat_customers_per_product'] = recom_df_with_aggs['repeat_customers_per_product'] / recom_df_with_aggs['unique_customers_per_product']


# Repeat products per customer
customer_product_counts = (
    recom_df_with_aggs
        .groupby(['Customer_ID', 'Product_ID'])
        .size()
)
repeat_products_per_customer = (
    customer_product_counts
        .loc[lambda s: s > 1]
        .reset_index(level='Product_ID')
        .groupby('Customer_ID')['Product_ID']
        .agg(list)
)
recom_df_with_aggs['repeat_products_per_customer'] = (
    recom_df_with_aggs['Customer_ID'].map(repeat_products_per_customer)
)

# Count of repeat products per customer
num_repeat_products_per_customer = (
    recom_df_with_aggs
        .groupby(['Customer_ID', 'Product_ID'])
        .size()
        .gt(1)
        .groupby('Customer_ID')
        .sum()
)
recom_df_with_aggs['num_repeat_products_per_customer'] = (
    recom_df_with_aggs['Customer_ID']
        .map(num_repeat_products_per_customer)
        .fillna(0)
        .astype(int)
)

# Pct of products repeated per customer
recom_df_with_aggs['pct_products_repeated_per_customer'] = recom_df_with_aggs['num_repeat_products_per_customer'] / recom_df_with_aggs['unique_products_per_customer']

# Count of repeat purchases per customer
repeat_purchases_per_customer = (
    recom_df_with_aggs
        .groupby(['Customer_ID', 'Product_ID'])
        .size()
        .sub(1)            # purchase_count - 1
        .clip(lower=0)     # no negatives
        .groupby('Customer_ID')
        .sum()
)
recom_df_with_aggs['num_repeat_purchases_per_customer'] = (
    recom_df_with_aggs['Customer_ID']
        .map(repeat_purchases_per_customer)
        .fillna(0)
        .astype(int)
)
```

# Recommendation System

```python
interactions = (
    transactions_df
      .groupby(['Customer_ID', 'Product_ID'])['Amount']
      .size()
      .reset_index(name='quantity_purchased')
)

from sklearn.metrics.pairwise import cosine_similarity

item_user = interactions.pivot_table(
    index='Product_ID',
    columns='Customer_ID',
    values='quantity_purchased',
    fill_value=0
)

item_similarity = cosine_similarity(item_user)

item_similarity_df = pd.DataFrame(
    item_similarity,
    index=item_user.index,
    columns=item_user.index
)

customer_products = (
    interactions
      .groupby('Customer_ID')['Product_ID']
      .apply(set)
)

def recommend_for_customer(customer_id, 
                           customer_products, 
                           item_similarity_df, 
                           n=10):

    purchased = customer_products.get(customer_id, set())

    if not purchased:
        return pd.Series(dtype=float)

    scores = pd.Series(0, index=item_similarity_df.index, dtype=float)

    for product in purchased:
        if product in item_similarity_df.index:
            scores += item_similarity_df[product]

    # Remove already purchased items
    scores = scores.drop(labels=purchased, errors='ignore')

    return scores.sort_values(ascending=False).head(n)

rand_cust = recom_df['Customer_ID'].sample().iloc[0]
recommend_for_customer(rand_cust, customer_products, item_similarity_df)
```

# Everything after here may be mental illness

```python
def compute_picky_repeater(row):
    if row['num_repeat_products_per_customer'] == 0:
        return 0
    depth = row['num_repeat_purchases_per_customer'] / row['num_repeat_products_per_customer']
    breadth_penalty = 1 - (row['num_repeat_products_per_customer'] / row['unique_products_per_customer'])
    return depth * breadth_penalty

customers_df['picky_repeater_score'] = customers_df.apply(compute_picky_repeater, axis=1)

#
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
customers_df['picky_repeater_score_norm'] = scaler.fit_transform(
    customers_df[['picky_repeater_score']]
)
```

```python
customers_df
```

```python
customers_df.nlargest(10, 'picky_repeater_score_norm')
```

```python
# Flag purchases above median
recom_df_with_aggs['premium_purchase'] = recom_df_with_aggs['price_vs_median'] > 0

# Frequency of premium purchases per customer
premium_freq = recom_df_with_aggs.groupby('Customer_ID')['premium_purchase'].mean()

# Average premium per customer (ignoring non-premium purchases)
avg_premium = recom_df_with_aggs[recom_df_with_aggs['premium_purchase']].groupby('Customer_ID')['price_vs_median'].mean()

# Combine into a single metric
willingness_to_pay_score = premium_freq * avg_premium

# Merge back into a customer-level dataframe
customers_df = customers_df.merge(
    willingness_to_pay_score.rename('willingness_to_pay_score'),
    left_on='Customer_ID',
    right_index=True,
    how='left'
)
```

```python
#
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
customers_df['willingness_to_pay_score_norm'] = scaler.fit_transform(
    customers_df[['willingness_to_pay_score']]
)
```
