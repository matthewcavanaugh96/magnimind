```python
import pandas as pd
import numpy as np

# Adjustable parameters for adjusted score
alpha, beta = 2, 2

# Load main player data
mlb_players = pd.read_csv('All Pitchers Cleaned.csv')

# Load model results
results_dtree = pd.read_csv('stored_results/Results_DTree_GS_Best_F1_Standard_KFold.csv')
results_etc = pd.read_csv('stored_results/Results_ETC_GS_Best_F1_Standard_KFold.csv')
results_logreg = pd.read_csv('stored_results/Results_Logreg_no_GS.csv')
results_rfc = pd.read_csv('stored_results/Results_RFC_GS_Best_F1_Standard_KFold.csv')
results_nn = pd.read_csv('stored_results/Results_Neural_Network_1.csv')

# Organize all models dynamically
model_dfs = {
    "Decision Tree": results_dtree,
    "Extra Trees": results_etc,
    "Logistic Regression": results_logreg,
    "Random Forest": results_rfc,
    "Neural Network": results_nn
}

# Store all player scores across models
player_scores = []

for model_name, df in model_dfs.items():
    # Skip models missing confidence info
    if 'Prediction Confidence' not in df.columns:
        print(f"⚠️ Skipping {model_name}: No 'Prediction Confidence' column found.")
        continue

    # Add 'Prediction Correct' if missing
    if 'Prediction Correct' not in df.columns and {'Hall of Fame', 'Prediction'}.issubset(df.columns):
        df['Prediction Correct'] = (df['Hall of Fame'] == df['Prediction'])

    # Calculate adjusted score
    df['Adjusted Prediction Score'] = np.where(
        df['Prediction Correct'],
        df['Prediction Confidence'] ** alpha,
        -beta * (df['Prediction Confidence'] ** alpha)
    )

    # Keep only relevant columns for aggregation
    model_data = df[['Player', 'Prediction Correct', 'Adjusted Prediction Score']].copy()
    model_data['Model'] = model_name
    player_scores.append(model_data)

# Combine results from all models
all_scores = pd.concat(player_scores, ignore_index=True)

# --- Calculate player-level metrics ---
player_metrics = (
    all_scores.groupby('Player')
    .agg(
        Accuracy_Percent=('Prediction Correct', lambda x: 100 * x.mean()),
        Avg_Adjusted_Score=('Adjusted Prediction Score', 'mean')
    )
    .reset_index()
)




# --- Merge metrics into main dataset ---
mlb_players = mlb_players.merge(player_metrics, on='Player', how='left')

mlb_players.rename(columns={'Accuracy_Percent': 'Percent of Predictions Accurate'}, inplace=True)
mlb_players.rename(columns={'Avg_Adjusted_Score': 'Mean Adjusted Prediction Score'}, inplace=True)


# --- (Optional) Save updated model files with adjusted scores ---
#for model_name, df in model_dfs.items():
#    out_path = f"stored_results/{model_name.replace(' ', '_')}_with_AdjustedScore.csv"
#    df.to_csv(out_path, index=False)

# Preview results
mlb_players
```

```python
results_comparison = pd.DataFrame()
results_comparison['Player'] = mlb_players['Player']
results_comparison['Hall of Fame'] = mlb_players['Hall of Fame']

results_comparison['DTree Prediction'] = results_dtree['Prediction']
results_comparison['DTree Correct'] = results_dtree['Prediction Correct']
results_comparison['DTree Confidence'] = results_dtree['Prediction Confidence']

results_comparison['ETC Prediction'] = results_etc['Prediction']
results_comparison['ETC Correct'] = results_etc['Prediction Correct']
results_comparison['ETC Confidence'] = results_etc['Prediction Confidence']

results_comparison['GBC Prediction'] = results_gbc['Prediction']
results_comparison['GBC Correct'] = results_gbc['Prediction Correct']
results_comparison['GBC Confidence'] = results_gbc['Prediction Confidence']

results_comparison['Logreg Prediction'] = results_logreg['Prediction']
results_comparison['Logreg Correct'] = results_logreg['Prediction Correct']
results_comparison['Logreg Confidence'] = results_logreg['Prediction Confidence']

results_comparison['RFC Prediction'] = results_rfc['Prediction']
results_comparison['RFC Correct'] = results_rfc['Prediction Correct']
results_comparison['RFC Confidence'] = results_rfc['Prediction Confidence']

results_comparison['SVC Prediction'] = results_svc['Prediction']
results_comparison['SVC Correct'] = results_svc['Prediction Correct']
results_comparison['SVC Confidence'] = results_svc['Prediction Confidence']





prediction_cols = ['DTree Prediction', 'ETC Prediction', 'GBC Prediction', 'Logreg Prediction', 'RFC Prediction', 'SVC Prediction']
results_comparison['Number of Positive Predictions'] = results_comparison[prediction_cols].eq(1).sum(axis=1)
results_comparison['Number of Negative Predictions'] = results_comparison[prediction_cols].eq(0).sum(axis=1)



#results_comparison['Number of Correct Predictions'] = results_comparison[columns_to_check].eq(1).sum(axis=1)


columns_to_check = ['DTree Correct', 'ETC Correct', 'GBC Correct', 'Logreg Correct', 'RFC Correct', 'SVC Correct']


# Calculate the sum of '1's in the specified columns for each row
count_of_ones = results_comparison[columns_to_check].sum(axis=1)

# Calculate the percentage
results_comparison['Percent of Predictions Correct'] = (count_of_ones / len(columns_to_check)) * 100




results_comparison['Mean Prediction Confidence'] = (
    results_comparison['DTree Confidence'] +
    results_comparison['ETC Confidence'] +
    results_comparison['GBC Confidence'] +
    results_comparison['Logreg Confidence'] +
    results_comparison['RFC Confidence'] +
    results_comparison['SVC Confidence']) / 6

results_comparison['Pct Correct Minus Confidence'] = results_comparison['Percent of Predictions Correct'] - results_comparison['Mean Prediction Confidence']

results_comparison
```
