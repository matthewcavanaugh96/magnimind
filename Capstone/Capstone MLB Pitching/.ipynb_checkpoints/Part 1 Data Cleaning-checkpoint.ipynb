This dataset is relatively clean to begin with. However, we will make small adjustments for clarity.

```python
import pandas as pd
```

```python
all_pitchers_df = pd.read_csv('Pitchers All min 1000 IP.txt')
all_pitchers_df
```

```python
all_pitchers_df[['Debut Age', 'Retirement Age']] = all_pitchers_df['Age'].str.split('-', expand=True).astype(int)
all_pitchers_df
```

```python
all_pitchers_df.columns
```

```python
# Innings Pitched rank, artifact of how Stathead displays query results.
all_pitchers_df.drop('Rk', axis=1, inplace=True)

all_pitchers_df.rename(columns={'IP': 'Innings Pitched'}, inplace=True)
all_pitchers_df.rename(columns={'From': 'Debut Year'}, inplace=True)
all_pitchers_df.rename(columns={'To': 'Retirement Year'}, inplace=True)
all_pitchers_df.rename(columns={'W': 'Wins'}, inplace=True)
all_pitchers_df.rename(columns={'L': 'Losses'}, inplace=True)
all_pitchers_df.rename(columns={'W-L%': 'Win Percentage'}, inplace=True)
all_pitchers_df.rename(columns={'Dec': 'Total Decisions'}, inplace=True)
all_pitchers_df.rename(columns={'ERA': 'Earned Run Average'}, inplace=True)
all_pitchers_df.rename(columns={'G': 'Total Games'}, inplace=True)
all_pitchers_df.rename(columns={'GS': 'Games Started'}, inplace=True)
all_pitchers_df.rename(columns={'CG': 'Complete Games'}, inplace=True)
all_pitchers_df.rename(columns={'SHO': 'Shutouts'}, inplace=True)
all_pitchers_df.rename(columns={'SV': 'Saves'}, inplace=True)

# Duplicate of Innings Pitched, artifact of how Stathead displays query results
all_pitchers_df.drop('IP.1', axis=1, inplace=True)

all_pitchers_df.rename(columns={'H': 'Hits Allowed'}, inplace=True)
all_pitchers_df.rename(columns={'R': 'Runs Allowed'}, inplace=True)
all_pitchers_df.rename(columns={'ER': 'Earned Runs Allowed'}, inplace=True)
all_pitchers_df.rename(columns={'HR': 'Home Runs Allowed'}, inplace=True)
all_pitchers_df.rename(columns={'BB': 'Walks Allowed'}, inplace=True)

# May drop this as it was not tracked until 1955.
all_pitchers_df.rename(columns={'IBB': 'Intentional Walks Allowed'}, inplace=True)

all_pitchers_df.rename(columns={'SO': 'Strikeouts'}, inplace=True)
all_pitchers_df.rename(columns={'HBP': 'Batters Hit By Pitch'}, inplace=True)
all_pitchers_df.rename(columns={'BK': 'Balks'}, inplace=True)
all_pitchers_df.rename(columns={'WP': 'Wild Pitches'}, inplace=True)

all_pitchers_df.rename(columns={'BP': 'Batters Faced'}, inplace=True)
#all_pitchers_df.rename(columns={'ERA+': 'Ballpark Adjusted ERA'}, inplace=True)
#all_pitchers_df.rename(columns={'FIP': 'Fielding Independent Pitching'}, inplace=True)
#all_pitchers_df.rename(columns={'WHIP': 'Walks Plus Hits Per Inning Pitched'}, inplace=True)
#all_pitchers_df.rename(columns={'H9': 'Hits Per 9 Innings Pitched'}, inplace=True)
#all_pitchers_df.rename(columns={'HR9': 'Year'}, inplace=True)

# Tracks other positions the pitcher may have played at some point. Unfair comparison, so dropping.
all_pitchers_df.drop('Pos', axis=1, inplace=True)

# Lists every MLB the pitcher played for. Unfair comparison, so dropping.
all_pitchers_df.drop('Team', axis=1, inplace=True)
```

```python
# New columns
all_pitchers_df[['Debut Age', 'Retirement Age']] = all_pitchers_df['Age'].str.split('-', expand=True).astype(int)
all_pitchers_df.drop('Age', axis=1, inplace=True)

all_pitchers_df['Relief Appearances'] = all_pitchers_df['Total Games'] - all_pitchers_df['Games Started']

all_pitchers_df['Career Length'] = all_pitchers_df['Retirement Year'] - all_pitchers_df['Debut Year']
```

```python
all_pitchers_df.columns
```

```python
all_pitchers_df = all_pitchers_df[['Player', 'Debut Year', 'Debut Age', 'Retirement Year', 'Retirement Age',
       'Career Length', 'Wins', 'Losses', 'Win Percentage', 'Total Decisions',
       'Earned Run Average', 'Total Games', 'Games Started', 'Relief Appearances', 'Complete Games',
       'Shutouts', 'Saves', 'Innings Pitched', 'Hits Allowed', 'Runs Allowed',
       'Earned Runs Allowed', 'Home Runs Allowed', 'Walks Allowed',
       'Intentional Walks Allowed', 'Strikeouts', 'Batters Hit By Pitch',
       'Balks', 'Wild Pitches', 'BF', 'ERA+', 'FIP', 'WHIP', 'H9', 'HR9',
       'BB9', 'SO9', 'SO/BB', 'Player-additional']]
```

```python
all_pitchers_df
```

```python
all_pitchers_df
```

```python
all_star_count = pd.read_csv('Pitchers All Star count.txt')
all_star_count.rename(columns={'Count': 'All Star Appearances'}, inplace=True)
all_star_count = all_star_count[['Player', 'Player-additional', 'All Star Appearances']]
all_star_count
```

```python
hall_of_fame_df = pd.read_csv('Pitchers HOF.txt')
hall_of_fame_df = hall_of_fame_df[['Player', 'Player-additional']]
hall_of_fame_df
```

Merging with HOF data.

```python
all_pitchers_df['Hall of Fame'] = all_pitchers_df['Player-additional'].isin(
    hall_of_fame_df['Player-additional']
).astype(int)
```

```python
all_pitchers_df
```

Merging with All Star count.

```python
# Merge the All-Star counts into all_pitchers_df
all_pitchers_df = all_pitchers_df.merge(
    all_star_count[['Player-additional', 'All Star Appearances']],  # ensure only needed columns
    on='Player-additional',
    how='left'
)

# Replace missing values (players with no All-Star appearances) with 0
all_pitchers_df['All Star Appearances'] = all_pitchers_df['All Star Appearances'].fillna(0).astype(int)
```

```python
all_pitchers_df
```

Resolve player name ambiguities

```python
all_pitchers_df['Player'].value_counts()
```

```python
# Identify player names that appear more than once
duplicates = all_pitchers_df['Player'].duplicated(keep=False)

# For those duplicates, append the unique code from 'Player-additional'
all_pitchers_df.loc[duplicates, 'Player'] = (
    all_pitchers_df.loc[duplicates, 'Player']
    + ' (' + all_pitchers_df.loc[duplicates, 'Player-additional'].astype(str) + ')'
)
```

```python
all_pitchers_df[all_pitchers_df['Player'].str.contains('Joe Coleman')]
```

```python
all_pitchers_df['Hall of Fame'].value_counts()
```

Nothing went wrong here, though I had to check. This is an artifact of occasional pitching appearances by non-pitchers, which usually occurs only in desperate situations. Thus, even though 124 players with at least one pitching appearance are in the Hall of Fame, only 82 of those met the 1000 innings requirement.

```python
all_pitchers_df['Player'].nunique()
```

```python
all_pitchers_df['Player-additional'].nunique()
```

We have successfully verified that all player name are disambiguated, so now we can drop the Player-additional column.

```python
all_pitchers_df.drop('Player-additional', axis=1, inplace=True)
```

The All-Star question?

The MLB All-Star game was not introduced until 1933. Some players in this dataset played in leagues which are retroactively recognized as major leagues, such as the Negro Leagues, and thus were not in MLB. Additionally, there are further non-statistical factors regarding All-Star selection. Because of this, I am dropping it here. However, its integration into the dataset further back in this notebook will remain, so that it can be re-included simply by commenting out the following cell.

```python
all_pitchers_df.drop('All Star Appearances', axis=1, inplace=True)
```

```python
all_pitchers_df.isnull().sum()
```

```python
all_pitchers_df = all_pitchers_df.fillna(0)
```

```python
all_pitchers_df.to_csv('All Pitchers Cleaned.csv', index=False)
```
