```python
def invert_features(df, columns):
    """
    Invert specified columns in a DataFrame so that higher values are more favorable.
    Simply multiplies values by -1. Operates directly on the original DataFrame.

    Parameters
    ----------
    df : pd.DataFrame
        The DataFrame to modify.
    columns : list of str
        Column names to invert.

    Returns
    -------
    None
        (modifies df in place)
    """
    for col in columns:
        if col in df.columns:
            df[col] = -df[col]
        else:
            print(f"Warning: Column '{col}' not found in DataFrame.")
```

```python
invert_features(mlb_pitchers, ['Losses', 'Earned Run Average', 'Hits Allowed', 'Runs Allowed', 'Earned Runs Allowed', 'Home Runs Allowed', 'Walks Allowed', 'Intentional Walks Allowed', 'Batters Hit By Pitch', 'Balks', 'Wild Pitches', 'FIP', 'WHIP', 'H9', 'HR9', 'BB9', 'Debut Age'])
```
