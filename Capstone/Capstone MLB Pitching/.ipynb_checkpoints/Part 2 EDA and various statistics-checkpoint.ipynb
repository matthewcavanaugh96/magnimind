```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
mlb_pitchers = pd.read_csv('All Pitchers Cleaned.csv')
mlb_pitchers
```

```python
mlb_pitchers.columns
```

```python
top20games = mlb_pitchers.sort_values('Total Games', ascending=False).head(20)
plt.figure(figsize=(10, 6))
plt.barh(top20games['Player'], top20games['Total Games'], color='green')
plt.gca().invert_yaxis()
plt.xlabel('Total Games')
plt.ylabel('Player')
plt.title('Top 20 MLB Pitchers by Total Games')
plt.tight_layout()
plt.show()
```

```python
top20starts = mlb_pitchers.sort_values('Games Started', ascending=False).head(20)
plt.figure(figsize=(10, 6))
plt.barh(top20starts['Player'], top20starts['Games Started'], color='green')
plt.gca().invert_yaxis()
plt.xlabel('Total Starts')
plt.ylabel('Player')
plt.title('Top 20 MLB Pitchers by Career Starts')
plt.tight_layout()
plt.show()
```

```python
top20so = mlb_pitchers.sort_values('Strikeouts', ascending=False).head(20)
plt.figure(figsize=(10, 6))
plt.barh(top20so['Player'], top20so['Strikeouts'], color='green')
plt.gca().invert_yaxis()
plt.xlabel('Total Games')
plt.ylabel('Player')
plt.title('Top 20 MLB Pitchers by Strikeouts')
plt.tight_layout()
plt.show()
```

```python
top20era = mlb_pitchers.sort_values('Earned Run Average', ascending=True).head(20)
plt.figure(figsize=(10, 6))
plt.barh(top20era['Player'], top20era['Earned Run Average'], color='green')
plt.gca().invert_yaxis()
plt.xlabel('Total Games')
plt.ylabel('Player')
plt.title('Top 20 MLB Pitchers by ERA (lower better)')
plt.tight_layout()
plt.show()
```

# Which statistics are most strongly correlated to Hall of Fame selection?

Since some stats (such as Earned Run Average) are ideally lower, we will sort by *absolute* correlation.

```python
mlb_pitchers_np = mlb_pitchers.drop('Player', axis=1)
correlations = mlb_pitchers_np.corrwith(mlb_pitchers_np['Hall of Fame'])

# Sort by absolute correlation
ranked_correlations = correlations.abs().sort_values(ascending=False)
ranked_correlations
```

This is a positive sign. I worried that relief pitchers would be punished for having fewer or no games started, and fewer career innings. Later on we will look deeper into whether this may be true, and indeed a high number of starts is rewarded, but for now it seems that a high number of Relief Appearances doesn't seem to have much impact.
