We will analyze every pitcher in MLB history that has at least 1000 innings pitched, and retired no later than 2019 (to account for the eligibility delay), and attempt to predict whether each one has been named to the Baseball Hall of Fame.

This dataset is highly imbalanced, with 82 Hall of Famers and 1133 non-Hall of Famers. 

We will use F1 score in Grid Searches. False positives and false negatives are equally important to avoid in this scenario, with no real-world harm to be done by any incorrect prediction.
