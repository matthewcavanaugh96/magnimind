```python
import pandas as pd
```

```python
df_job_uniqueness = pd.read_csv('transformed data/2-Job Uniqueness by Tech Tasks and Tools.csv')
df_job_uniqueness
```

```python
df_t3_uniqueness = df_job_uniqueness.copy()
df_t3_uniqueness['Mean T3 Uniqueness'] = (df_t3_uniqueness['AVG_TECH_SKILL_UNIQUENESS_1to5'] + df_t3_uniqueness['AVG_TOOL_UNIQUENESS_1to5'] +df_t3_uniqueness['AVG_TASK_UNIQUENESS_1to5']) / 3
df_t3_uniqueness = df_t3_uniqueness[['O*NET-SOC Code', 'Job Title', 'Mean T3 Uniqueness']]
df_t3_uniqueness
```

```python
df_t3_uniqueness.nsmallest(10, 'Mean T3 Uniqueness')
```

```python
df_scaled_elements = pd.read_csv('transformed data/2-Prescaled Elements normalized comb level and import.csv')

# Have to go back in later and implement this change in the original notebook
df_scaled_elements.rename(columns={'Title': 'Job Title'}, inplace=True)

df_scaled_elements
```

```python
excluded_cols = ['O*NET-SOC Code', 'Job Title']
df_scaled_elems_avg = df_scaled_elements.copy()
df_scaled_elems_avg['Element Avg'] = df_scaled_elems_avg.drop(columns=excluded_cols).mean(axis=1)
df_scaled_elems_avg = df_scaled_elems_avg[['O*NET-SOC Code', 'Job Title', 'Element Avg']]
df_scaled_elems_avg
```

```python
df_scaled_elems_avg.nsmallest(10, 'Element Avg')
```
