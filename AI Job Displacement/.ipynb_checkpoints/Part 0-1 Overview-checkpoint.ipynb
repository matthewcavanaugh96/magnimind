# My attributes and scoring

1. Routine structure - How repetitive is the job, how predictable are the tasks?

2. Cognitive complexity - How advanced is the reasoning required by this job? Do unique problems arise which cannot be solved with simple rules?

3. Physical requirements - How much physical activity is required? Lifting, manual adjustments, etc.

4. Social interactions - How frequently does the job require interactions with people? How important are these interactions to job success? How complex are these interactions i.e. do they require more depth than, for instance, a phonebot that forces selection from a narrow list to hear pre-recorded answers?

5. Creativity - How much is originality incentivized over following instructions?

6. Decision accountability - How crucial are the decisions being made, and to what extent is it preferred that humans have final say-so above automated processes?

7. Data structure - How readily available are structured data and objective performance metrics?

Each of these will be normalized for directionality.

Dimension Weights

Routine - 20% - Positive impact on score

Cognitive complexity - 15% - Negative impact on score

Physical requirements - 15% - Negative impact on score

Social interactions - 15% - Negative impact on score

Creativity - 15% - Negative impact on score

Decision accountability 10% - Negative impact on score

Data structure 10% - Positive impact on score

Several sheets within the O*NET dataset already had pre-scaled Elements - these encompass skills, abilities, work activities, knowledge, and work context (situations). I pivoted the data, normalized all to fit a 1-5 scale, and balanced importance with level for the elements that had both. 

I asked AI models to assign each of my transformed elements into one of the seven attributes listed above. I calculated an average for each attribute, then weighed them according to my scale. This was done for three models so far - ChatGPT, Gemini, and Claude (more may be added later). Finally, I took an average of each model's score to gain a broader consensus.

Parts 1 and 2 contain various stages of data cleaning.

Part 3 is where my final dataset, using in my .py page, is put together.
