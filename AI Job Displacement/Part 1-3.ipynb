How many titles is each task associated with?

```python
titles_per_task = (
    df_task_statements
        .groupby("Task")["Title"]
        .nunique()
        .reset_index(name="Number of Titles")
)

titles_per_task
```

```python
titles_per_task["Uniqueness Rank DENSE"] = (
    titles_per_task["Number of Titles"]
        .rank(method="dense", ascending=True)
)

titles_per_task = titles_per_task.sort_values("Number of Titles")

titles_per_task["Uniqueness Rank MIN"] = (
    titles_per_task["Number of Titles"]
        .rank(method="min", ascending=True)
        .astype(int)
)



titles_per_task
```
