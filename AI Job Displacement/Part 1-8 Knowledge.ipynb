```python
import pandas as pd
df_technology_skills = pd.read_csv('ONET data/Technology Skills.csv')
df_tools_used = pd.read_csv('ONET data/Tools Used.csv')
df_task_statements = pd.read_csv('ONET data/Task Statements.csv')
df_skills = pd.read_csv('ONET data/Skills.csv')
df_abilities = pd.read_csv('ONET data/Abilities.csv')
df_occupation_data = pd.read_csv('ONET data/Occupation Data.csv')
df_work_activities = pd.read_csv('ONET data/Work Activities.csv')

df_knowledge = pd.read_csv('ONET data/Knowledge.csv')
df_work_context = pd.read_csv('ONET data/Work Context.csv')
```

```python
df = df_work_activities.copy()

# Create combined column label
df["Skill_Scale"] = df["Element Name"] + " " + df["Scale Name"]
```

```python
df_wide = df.pivot_table(
    index=["O*NET-SOC Code", "Title"],
    columns="Skill_Scale",
    values="Data Value"
).reset_index()
```

```python
df_wide
```

```python
df_wide.to_csv('transformed data/8-Knowledge Pivoted.csv', index=False)
```

# Z-Score

```python
df_z = df_wide.copy()

# Identify numeric skill columns
skill_cols = df_wide.select_dtypes(include="number").columns


for col in skill_cols:
    df_z[col] = (
        (df_wide[col] - df_wide[col].mean()) /
        df_wide[col].std()
    )
```

```python
meta_cols = ["O*NET-SOC Code", "Title"]

level_cols = [
    col for col in df_z.columns
    if "Level" in col
]

import_cols = [
    col for col in df_z.columns
    if "Importance" in col
]
```

```python
df_z
```

```python
df_z.to_csv('transformed data/8-Knowledge Z-Scores.csv', index=False)
```

# Z Score Aggregates

```python
df_z_aggs = df_z.copy()



df_z_aggs["z_level_mean"] = df_z_aggs[level_cols].mean(axis=1)
df_z_aggs["z_level_median"] = df_z_aggs[level_cols].median(axis=1)
df_z_aggs["z_level_max"] = df_z_aggs[level_cols].max(axis=1)
df_z_aggs["z_level_min"] = df_z_aggs[level_cols].min(axis=1)
df_z_aggs["z_level_min+max"] = df_z_aggs["z_level_min"] + df_z_aggs["z_level_max"]

df_z_aggs["z_import_mean"] = df_z_aggs[import_cols].mean(axis=1)
df_z_aggs["z_import_median"] = df_z_aggs[import_cols].median(axis=1)
df_z_aggs["z_import_max"] = df_z_aggs[import_cols].max(axis=1)
df_z_aggs["z_import_min"] = df_z_aggs[import_cols].min(axis=1)
df_z_aggs["z_import_min+max"] = df_z_aggs["z_import_min"] + df_z_aggs["z_import_max"]

# Drop level_cols and import_cols as defined earlier
df_z_aggs = df_z_aggs.drop(columns=level_cols)
df_z_aggs = df_z_aggs.drop(columns=import_cols)
```

```python
df_z_aggs
```

```python
# Define columns to prefix
cols_to_prefix = df_z_aggs.drop(columns=['O*NET-SOC Code', 'Title']).columns
# Apply prefix to these columns
df_z_aggs_pref = df_z_aggs.rename(columns={c: f'KNOWLEDGE_{c}' for c in cols_to_prefix})
df_z_aggs_pref
```

```python
df_z_aggs_pref.to_csv('transformed data/8-Knowledge Z-Score Aggregates.csv', index=False)
```
