# Aggregates

```python
# Reset DF definition
df_tools_used = pd.read_csv('ONET data/Tools Used.csv')
df_tools_used['Commodity Code'] = df_tools_used['Commodity Code'].replace(23151807, 41104701)
```

```python
df_tools_used
```

```python
# Find count of unique tools by title
total_tools_by_title = (
    df_tools_used
    .groupby(['O*NET-SOC Code', 'Title'])['Example']
    .nunique()
    .reset_index(name='TOTAL_TOOLS_BY_JOB')
    .sort_values('TOTAL_TOOLS_BY_JOB', ascending=False)
)

# Find count of unique tool types by title
total_tool_types_by_title = (
    df_tools_used
    .groupby(['Title'])['Commodity Title']
    .nunique()
    .reset_index(name='TOTAL_TOOL_TYPES_BY_JOB')
    .sort_values('TOTAL_TOOL_TYPES_BY_JOB', ascending=False)
)

# Create column list of all tools by title
tools_by_title = (
    df_tools_used
    .groupby('Title')['Example']
    .apply(lambda x: '; '.join(
        f'"{e}"' for e in sorted(x.unique())
    ))
    .reset_index(name='TOOLS_LIST')
)

# Create column list of all tool types by title
tool_types_by_title = (
    df_tools_used
    .groupby('Title')['Commodity Title']
    .apply(lambda x: '; '.join(
        f'"{e}"' for e in sorted(x.unique())
    ))
    .reset_index(name='TOOL_TYPES_LIST')
)



```

```python
tool_types_by_title
```

```python
# Merge
titles_tools_and_types = df_tools_used.merge(
    total_tools_by_title,
    on=['Title', 'O*NET-SOC Code'],
    how='left'
)

# Merge
titles_tools_and_types = titles_tools_and_types.merge(
    total_tool_types_by_title,
    on=['Title'],
    how='left'
)

# Merge
titles_tools_and_types = titles_tools_and_types.merge(
    tools_by_title,
    on=['Title'],
    how='left'
)

# Merge
titles_tools_and_types = titles_tools_and_types.merge(
    tool_types_by_title,
    on=['Title'],
    how='left'
)
```

```python
titles_tools_and_types
```

```python
titles_tools_and_types.rename(columns={'Title': 'Job Title'}, inplace=True)
titles_tools_and_types.rename(columns={'Example': 'Tool Name'}, inplace=True)
titles_tools_and_types.rename(columns={'Commodity Code': 'Tool Type Code'}, inplace=True)
titles_tools_and_types.rename(columns={'Commodity Title': 'Tool Type Title'}, inplace=True)
```

```python
titles_tools_and_types
```

```python
titles_tools_and_types.to_csv('transformed data/2-Jobs Agg Tools.csv', index=False)
```

# Next - Need to aggregate Tools by Jobs - The reverse of what was done above

```python
# Reset DF definition
df_tools_used = pd.read_csv('ONET data/Tools Used.csv')
df_tools_used['Commodity Code'] = df_tools_used['Commodity Code'].replace(23151807, 41104701)
```

```python
# Find count of unique tools by title
total_tools_by_title = (
    df_tools_used
    .groupby(['O*NET-SOC Code', 'Title'])['Example']
    .nunique()
    .reset_index(name='TOTAL_TOOLS_BY_JOB')
    .sort_values('TOTAL_TOOLS_BY_JOB', ascending=False)
)

# Find count of unique tool types by title
total_tool_types_by_title = (
    df_tools_used
    .groupby(['Title'])['Commodity Title']
    .nunique()
    .reset_index(name='TOTAL_TOOL_TYPES_BY_JOB')
    .sort_values('TOTAL_TOOL_TYPES_BY_JOB', ascending=False)
)

# Create column list of all tools by title
tools_by_title = (
    df_tools_used
    .groupby('Title')['Example']
    .apply(lambda x: '; '.join(
        f'"{e}"' for e in sorted(x.unique())
    ))
    .reset_index(name='TOOLS_LIST')
)

# Create column list of all tool types by title
tool_types_by_title = (
    df_tools_used
    .groupby('Title')['Commodity Title']
    .apply(lambda x: '; '.join(
        f'"{e}"' for e in sorted(x.unique())
    ))
    .reset_index(name='TOOL_TYPES_LIST')
)



```

```python
# Find count of unique titles by tool
total_titles_by_tool = (
    df_tools_used
    .groupby(['Example'])['Title']
    .nunique()
    .reset_index(name='TOTAL_JOBS_BY_TOOL')
    .sort_values('TOTAL_JOBS_BY_TOOL', ascending=False)
)

# Find count of unique tools by title
total_titles_by_tool_type = (
    df_tools_used
    .groupby(['Commodity Title'])['Title']
    .nunique()
    .reset_index(name='TOTAL_JOBS_BY_TOOL_TYPE')
    .sort_values('TOTAL_JOBS_BY_TOOL_TYPE', ascending=False)
)

# Create column list of all job titles by tool
tools_by_title = (
    df_tools_used
    .groupby('Example')['Title']
    .apply(lambda x: '; '.join(
        f'"{e}"' for e in sorted(x.unique())
    ))
    .reset_index(name='JOBS_LIST')
)

# Create column list of all job titles by tool type
tool_types_by_title = (
    df_tools_used
    .groupby('Commodity Title')['Title']
    .apply(lambda x: '; '.join(
        f'"{e}"' for e in sorted(x.unique())
    ))
    .reset_index(name='TOOL_TYPES_LIST')
)



```

```python
# Merge
tools_agg_jobs = total_titles_by_tool.merge(
    tools_by_title,
    on=['Example'],
    how='left'
)

tools_agg_jobs
```

```python
#tools_agg_jobs.rename(columns={'Example': 'Tool Name'}, inplace=True)
tools_agg_jobs.to_csv('transformed data/2-Tools Agg Jobs.csv', index=False)
```

```python
# Merge
tool_types_agg_jobs = total_titles_by_tool_type.merge(
    tool_types_by_title,
    on=['Commodity Title'],
    how='left'
)

tool_types_agg_jobs
```

```python
tool_types_agg_jobs.rename(columns={'Commodity Title': 'Tool Type'}, inplace=True)
tool_types_agg_jobs.to_csv('transformed data/2-Tool Types Agg Jobs.csv', index=False)
```

```python
# Merge
titles_tools_and_types = df_tools_used.merge(
    total_tools_by_title,
    on=['Title', 'O*NET-SOC Code'],
    how='left'
)

# Merge
titles_tools_and_types = titles_tools_and_types.merge(
    total_tool_types_by_title,
    on=['Commodity Title', 'O*NET-SOC Code'],
    how='left'
)

# Merge
titles_tools_and_types = titles_tools_and_types.merge(
    tools_by_title,
    on=['Title'],
    how='left'
)

# Merge
titles_tools_and_types = titles_tools_and_types.merge(
    tool_types_by_title,
    on=['Title'],
    how='left'
)
```
