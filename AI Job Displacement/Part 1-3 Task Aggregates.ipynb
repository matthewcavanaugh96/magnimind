```python
df_task_statements['Task_clean'] = (
    df_task_statements['Task']
    .str.strip()
    .str.replace(r'\s+', ' ', regex=True)
)
```

```python
df_task_statements['Task_Concept_ID'] = (
    df_task_statements['Task_clean']
    .astype('category')
    .cat.codes
)
```

```python
df_task_statements['Concept_Instance_Number'] = (
    df_task_statements
    .groupby('Task_Concept_ID')
    .cumcount() + 1
)
```

```python
df_task_statements['Semantic_Instance_ID'] = (
    df_task_statements['Task_Concept_ID'].astype(str)
    + '-'
    + df_task_statements['Concept_Instance_Number'].astype(str)
)
```

# Now with unique identifiers, aggregates can be made properly.

```python
# Find count of unique task by title
total_tasks_by_title = (
    df_task_statements
    .groupby(['O*NET-SOC Code', 'Title'])['Task_String_ID']
    .nunique()
    .reset_index(name='TOTAL_TASKS_BY_JOB')
    .sort_values('TOTAL_TASKS_BY_JOB', ascending=False)
)
```

```python
total_tasks_by_title
```

```python
# Create column list of all tasks by title
tasks_by_title = (
    df_task_statements
    .groupby('Title')['Task']
    .apply(lambda x: '; '.join(
        f'"{e}"' for e in sorted(x.unique())
    ))
    .reset_index(name='TASK_LIST')
)
```

```python
tasks_by_title
```

```python
# Merge
tasks_and_concepts_by_job = total_tasks_by_title.merge(
    tasks_by_title,
    on=['Title'],
    how='left'
)

#tasks_and_concepts_by_job = tasks_and_concepts_by_job.drop('TOTAL_TASK_CONCEPTS_BY_JOB', axis=1)
tasks_and_concepts_by_job
```

```python
tasks_and_concepts_by_job.to_csv('transformed data/3-Jobs Agg Tasks.csv', index=False)
```

# Now jobs by task

```python
# Find count of unique titles by task
total_tasks_by_title = (
    df_task_statements
    .groupby(['Task_String_ID'])['Title']
    .nunique()
    .reset_index(name='TOTAL_JOBS_BY_TASK')
    .sort_values('TOTAL_JOBS_BY_TASK', ascending=False)
)
```

```python
# Create column list of all titles by task
tasks_by_title = (
    df_task_statements
    .groupby('Task_Concept_ID')['Title']
    .apply(lambda x: '; '.join(
        f'"{e}"' for e in sorted(x.unique())
    ))
    .reset_index(name='JOB_LIST')
)
```

```python
total_tasks_by_title
```

```python
tasks_by_title
```

```python
df_task_statements_big = df_task_statements.merge(
    total_tasks_by_title,
    on=['Task_String_ID'],
    how='left'
)
```

```python
df_task_statements_big = df_task_statements_big.merge(
    tasks_by_title,
    on=['Task_Concept_ID'],
    how='left'
)
```
