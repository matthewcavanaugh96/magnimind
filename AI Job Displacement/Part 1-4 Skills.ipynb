# Pivot

```python
df = df_skills.copy()

# Create combined column label
df["Skill_Scale"] = df["Element Name"] + " " + df["Scale Name"]
```

```python
df_wide = df.pivot_table(
    index=["O*NET-SOC Code", "Title"],
    columns="Skill_Scale",
    values="Data Value"
).reset_index()
```

```python
df_wide
```

```python
df_wide.to_csv('transformed data/4-Skills Pivoted.csv', index=False)
```

# Remove irrelevant skills 

```python
df_relevant = df_skills.copy()

# Rename column, relabel data, change column data type
df_relevant.rename(columns={'Not Relevant': 'Relevant'}, inplace=True)
df_relevant['Relevant'] = df_relevant['Relevant'].replace('Y', '0')
df_relevant['Relevant'] = df_relevant['Relevant'].replace('N', '1')
df_relevant['Relevant'] = df_relevant['Relevant'].fillna(1)
df_relevant['Relevant'] = df_relevant['Relevant'].astype('int')

# 
df_relevant['Recommend Suppress'] = df_relevant['Recommend Suppress'].replace('Y', '1')
df_relevant['Recommend Suppress'] = df_relevant['Recommend Suppress'].replace('N', '0')
df_relevant['Recommend Suppress'] = df_relevant['Recommend Suppress'].astype('int')
```

```python
df_relevant
```

```python
df_irrev = df_relevant[df_relevant['Recommend Suppress'] == 1]
df_irrev = df_irrev[df_irrev['Relevant'] == 0]
df_irrev
```

```python
df_rev_sup = df_relevant[df_relevant['Recommend Suppress'] == 1]
df_rev_sup = df_rev_sup[df_rev_sup['Relevant'] == 1]
df_rev_sup
```

```python
df_relevant.shape
```

```python
df_relevant_only = df_relevant[df_relevant['Recommend Suppress'] == 0]
df_relevant_only = df_relevant_only[df_relevant_only['Relevant'] == 1]
df_relevant_only
```

```python
# Create combined column label
df_relevant_only["Skill_Scale"] = df_relevant_only["Element Name"] + " " + df_relevant_only["Scale Name"]
```

```python
df_wide_rel = df_relevant_only.pivot_table(
    index=["O*NET-SOC Code", "Title"],
    columns="Skill_Scale",
    values="Data Value"
).reset_index()
```
