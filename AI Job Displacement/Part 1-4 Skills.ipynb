# Pivot

```python
df = df_skills.copy()

# Create combined column label
df["Skill_Scale"] = df["Element Name"] + " " + df["Scale Name"]
```

```python
df_wide = df.pivot_table(
    index=["O*NET-SOC Code", "Title"],
    columns="Skill_Scale",
    values="Data Value"
).reset_index()
```

```python
df_wide
```

```python
df_wide.to_csv('transformed data/4-Skills Pivoted.csv', index=False)
```

# Compare to average

```python
df_wide = df_wide.copy()

# Identify numeric skill columns
skill_cols = df_wide.select_dtypes(include="number").columns
```

```python
df_diff = df_wide.copy()

for col in skill_cols:
    df_diff[col] = df_wide[col] - df_wide[col].mean()
```

```python
df_diff
```

# Percent of mean

```python
df_pct = df_wide.copy()

# Identify numeric skill columns
skill_cols = df_wide.select_dtypes(include="number").columns

for col in skill_cols:
    mean_val = df_wide[col].mean()
    df_pct[col] = (df_wide[col] - mean_val) / mean_val * 100
```

```python
df_pct
```

```python
meta_cols = ["O*NET-SOC Code", "Title"]

level_cols = [
    col for col in df_pct.columns
    if "Level" in col
]

import_cols = [
    col for col in df_pct.columns
    if "Importance" in col
]

df_pct["z_level_avg"] = df_pct[level_cols].mean(axis=1)
df_pct["z_import_avg"] = df_pct[import_cols].mean(axis=1)
```

# Z Scores

```python
df_z = df_wide.copy()

# Identify numeric skill columns
skill_cols = df_wide.select_dtypes(include="number").columns


for col in skill_cols:
    df_z[col] = (
        (df_wide[col] - df_wide[col].mean()) /
        df_wide[col].std()
    )
```

# Z Score aggregates

```python
df_z_aggs = df_z.copy()



df_z_aggs["z_level_mean"] = df_z_aggs[level_cols].mean(axis=1)
df_z_aggs["z_level_median"] = df_z_aggs[level_cols].median(axis=1)
df_z_aggs["z_level_max"] = df_z_aggs[level_cols].max(axis=1)
df_z_aggs["z_level_min"] = df_z_aggs[level_cols].min(axis=1)
df_z_aggs["z_level_min+max"] = df_z_aggs["z_level_min"] + df_z_aggs["z_level_max"]

df_z_aggs["z_import_mean"] = df_z_aggs[import_cols].mean(axis=1)
df_z_aggs["z_import_median"] = df_z_aggs[import_cols].median(axis=1)
df_z_aggs["z_import_max"] = df_z_aggs[import_cols].max(axis=1)
df_z_aggs["z_import_min"] = df_z_aggs[import_cols].min(axis=1)
df_z_aggs["z_import_min+max"] = df_z_aggs["z_import_min"] + df_z_aggs["z_import_max"]

# Drop level_cols and import_cols as defined earlier
df_z_aggs = df_z_aggs.drop(columns=level_cols)
df_z_aggs = df_z_aggs.drop(columns=import_cols)
```
