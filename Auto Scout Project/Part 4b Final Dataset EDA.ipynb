```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
scout_car = pd.read_csv('Step 4 - Dummies created.csv')
scout_car
```

# Correlation heatmap

```python
#sns.set(style="whitegrid")

# Generate a mask for the upper triangle
mask = np.zeros_like(scout_car.corr(numeric_only = True), dtype=bool)
mask[np.triu_indices_from(mask)] = True

# Set up the matplotlib figure
f, ax = plt.subplots(figsize=(18, 18))

# Draw the heatmap with the mask and correct aspect ratio
sns.heatmap(scout_car.corr(numeric_only = True), mask=mask, cmap='Reds', vmax=1, center=0,
            square=True, linewidths=.5, cbar_kws={"shrink": .5}, annot=True)

plt.title("Correlation matrix for climate and demograhpic data", fontsize=1, fontweight='bold')
#plt.savefig("Correlation matrix for scout_car data.png")
plt.show()
```
