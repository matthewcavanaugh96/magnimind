```python
import pandas as pd
```

```python
scout_car = pd.read_csv('Step 3 - Outliers handled.csv')
scout_car
```

```python
scout_car.drop('Make', axis=1, inplace=True)
scout_car.drop('Model', axis=1, inplace=True)
```

```python
scout_car['Body Color'].value_counts()
```

In validation, 'Body Color_Black' should not be a column as it should be dropped.

```python
# Create dummy variables for all non-numeric columns and drop the originals
scout_car = pd.get_dummies(scout_car, drop_first=True)
```

```python
scout_car.shape
```

```python
bool_cols = scout_car.select_dtypes(include='bool').columns
scout_car[bool_cols] = scout_car[bool_cols].astype(int)
```

Fuel Consumption?

```python
scout_car[scout_car['Consumption (L /100 km) comb'] != scout_car['Consumption (L /100 km) country']]
```

```python
scout_car[scout_car['Consumption (L /100 km) comb'] != scout_car['Consumption (L /100 km) city']]
```

```python
scout_car[scout_car['Consumption (L /100 km) country'] != scout_car['Consumption (L /100 km) city']]
```

```python
scout_car.drop('Consumption (L /100 km) country', axis=1, inplace=True)
scout_car.drop('Consumption (L /100 km) city', axis=1, inplace=True)
```
