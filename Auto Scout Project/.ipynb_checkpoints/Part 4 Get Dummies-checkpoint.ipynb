Earlier I thought it could be useful to have 'Make' and 'Model' as independent columns, but now I have realized it will create many redundant dummy columns.

```python
scout_car.drop('Make', axis=1, inplace=True)
scout_car.drop('Model', axis=1, inplace=True)
```

```python
# Compute proportion of identical values (max frequency / total count)
identical_ratio = scout_car.apply(lambda col: col.value_counts(normalize=True).max())

# Sort columns by this percentage
sorted_cols = identical_ratio.sort_values(ascending=False)

pd.set_option('display.max_rows', None)  

sorted_cols
```

All three of these columns are entirely identical, so we can drop two of them.

```python
scout_car.drop('Consumption (L /100 km) country', axis=1, inplace=True)
scout_car.drop('Consumption (L /100 km) city', axis=1, inplace=True)
```

In validation after dummy creation, 'Body Color_Black' should not be a column as it should be dropped.

```python
# Create dummy variables for all non-numeric columns and drop the originals
scout_car = pd.get_dummies(scout_car, drop_first=True)

# Convert Boolean columns to integers with 1s and 0s
bool_cols = scout_car.select_dtypes(include='bool').columns
scout_car[bool_cols] = scout_car[bool_cols].astype(int)
```
