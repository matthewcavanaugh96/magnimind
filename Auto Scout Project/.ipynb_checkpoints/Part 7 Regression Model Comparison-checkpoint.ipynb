```python
results_rfr = pd.read_csv('/Users/matthewcavanaugh/Desktop/GitHub/magnimind/Auto Scout Project/Model Results/RESULTS RFR no GS')

# Calculating R2 Score manually from results Dataframe
y_rfr = results_rfr['Actual Price']
y_pred_rfr = results_rfr['Predicted Price']

ss_res = np.sum((y_rfr - y_pred_rfr) ** 2)           # residual sum of squares
ss_tot = np.sum((y_rfr - y_pred_rfr.mean()) ** 2)    # total sum of squares

r2 = 1 - (ss_res / ss_tot)
print("R² Score:", r2)
```

```python
results_gbr = pd.read_csv('/Users/matthewcavanaugh/Desktop/GitHub/magnimind/Auto Scout Project/Model Results/RESULTS GBR no GS')

# Calculating R2 Score manually from results Dataframe
y_gbr = results_gbr['Actual Price']
y_pred_gbr = results_gbr['Predicted Price']

ss_res = np.sum((y_gbr - y_pred_gbr) ** 2)           # residual sum of squares
ss_tot = np.sum((y_gbr - y_pred_gbr.mean()) ** 2)    # total sum of squares

r2 = 1 - (ss_res / ss_tot)
print("R² Score:", r2)
```
