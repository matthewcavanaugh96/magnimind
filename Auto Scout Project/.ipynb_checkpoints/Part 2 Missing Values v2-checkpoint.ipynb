# Missing value filling

```python
import pandas as pd
import numpy as np
```

```python
scout_car = pd.read_csv('Step 1 - Cleaned v2.csv')

# Remove leading/trailing spaces and newlines
scout_car = scout_car.applymap(lambda x: x.strip() if isinstance(x, str) else x)

scout_car
```

```python
scout_car.isnull().sum()[scout_car.isnull().sum() > 0]
```

```python
scout_car.dtypes
```

# Columns not necessary or irrelevant in machine learning?

These columns are complicated strings.

'URL'

'Short description'

'Offer number'

'Description' - But maybe I could use sentiment analysis and assign a numeric score accordingly? Would have to translate though, how reliable would translations be? I'll drop it for now but may come back to reincorporate it later.

```python
scout_car.drop('URL', axis=1, inplace=True)
scout_car.drop('Short Description', axis=1, inplace=True)
scout_car.drop('Offer Number', axis=1, inplace=True)
```

# Null Fillings

```python
scout_car['Body Type'].value_counts(dropna=False)
```

```python
# Fill Nulls with the most common value based on Make_Model
scout_car["Body Type"] = (
    scout_car.groupby("make_model")["Body Type"]
             .transform(lambda x: x.fillna(x.mode().iloc[0] if not x.mode().empty else x))
)
```

```python
scout_car['Price'].isnull().sum()
```

```python
# Fill Nulls with the most common value based on Make_Model
scout_car["VAT"] = (
    scout_car.groupby("make_model")["VAT"]
             .transform(lambda x: x.fillna(x.mode().iloc[0] if not x.mode().empty else x))
)
```

'Registration YYYYMM' should be changed to YYYY only.

```python
# Use integer division to remove the last two digits
scout_car['Registration (YYYYMM)'] = scout_car['Registration (YYYYMM)'] // 100
scout_car.rename(columns={'Registration (YYYYMM)': 'Registration'}, inplace=True)
scout_car['Registration'].isnull().sum()
```

```python
# Fill Nulls with the most common value based on make_model
scout_car["Registration"] = (
    scout_car.groupby("make_model")["Registration"]
             .transform(lambda x: x.fillna(x.mode().iloc[0] if not x.mode().empty else x))
)
```

All 34 Renault Dusters, and the one Audi A2, had no values for weight to begin with. Each also lacks certain other values, thus there was no median to calculate when we did so earlier. Let's drop rows with these types.

```python
models_to_drop = ['Audi A2', 'Renault Duster']
scout_car = scout_car[~scout_car['make_model'].isin(models_to_drop)]
```

```python
null_counts = scout_car.isnull().sum().sort_values(ascending=False)
print(null_counts)
```
