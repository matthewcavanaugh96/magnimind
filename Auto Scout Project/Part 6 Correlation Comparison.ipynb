# Correlation Comparison

```python
import pandas as pd
```

```python
linreg_corr = pd.read_csv('/Users/matthewcavanaugh/Desktop/GitHub/magnimind/Auto Scout Project/Model Results/Correlation Rankings LinReg.csv')
logreg_corr = pd.read_csv('/Users/matthewcavanaugh/Desktop/GitHub/magnimind/Auto Scout Project/Model Results/Correlation Rankings LogReg.csv')
corr_comp = pd.DataFrame()
```

```python
corr_comp = pd.merge(
    linreg_corr[['Column', 'AbsCorrelation Rank']],
    logreg_corr[['Column', 'AbsCorrelation Rank']],
    on='Column',
    how='outer',  # use 'outer' if some columns exist only in one file, else 'inner'
    suffixes=('_LinReg', '_LogReg')
)

# Rename for clarity
corr_comp.rename(
    columns={
        'AbsCorrelation Rank_LinReg': 'LinReg Corr Rank',
        'AbsCorrelation Rank_LogReg': 'LogReg Corr Rank'
    },
    inplace=True
)

corr_comp['Average Corr Rank'] = (corr_comp['LinReg Corr Rank'] + corr_comp['LogReg Corr Rank']) / 2

corr_comp['Difference'] = (corr_comp['LinReg Corr Rank'] - corr_comp['LogReg Corr Rank']).abs()

corr_comp['RPD'] = corr_comp['Average Corr Rank'] + corr_comp['Difference']

pd.set_option('display.max_rows', None)

corr_comp.sort_values(by='RPD', ascending=True)
```
