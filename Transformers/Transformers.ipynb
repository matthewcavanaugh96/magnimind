In deep learning, the transformer is a neural network architecture based on the multi-head attention mechanism, in which text is converted to numerical representations called tokens, and each token is converted into a vector via lookup from a word embedding table.[1] At each layer, each token is then contextualized within the scope of the context window with other (unmasked) tokens via a parallel multi-head attention mechanism, allowing the signal for key tokens to be amplified and less important tokens to be diminished.

A transformer is a kind of machine learning model. It is a neural network that repeats modules with duplicated parameters like a convolutional network, but instead of using receptive fields of fixed connections to spread information horizontally, it uses a technique called attention. The way attention works, is that each module outputs not just a value vector, but also a key vector and a query vector for the corresponding module in the next layer. In the next layer, the input to a module is generally a weighted sum of value vectors from all modules in the preceding layer, weighted by how well the query vector matches the key vectors from the preceding layer’s modules.

A key vector says “I’ve got this kind of information.” A query vector says “I’m interested in this kind of information.” A value vector is just a value. Most models use multiple “attention heads,” meaning they use multiple sets of queries, keys, and values to connect module layers.

🧠 What Are Transformers?

Transformers are a type of neural network architecture introduced in the 2017 paper “Attention Is All You Need” by Vaswani et al. They revolutionized how machines process sequential data (like text, audio, and time series) by replacing traditional sequence models such as RNNs and LSTMs.

⚙️ Core Idea: Attention Mechanism

Instead of processing data sequentially, transformers process all tokens in parallel and use self-attention to decide which parts of the input are most relevant to each other.

In simple terms:
Each word (or token) looks at every other word to decide how much attention it should give them when forming its representation.

Example (in a sentence):

In the sentence “The cat sat on the mat,”
the word “cat” will attend strongly to “sat”, because those are closely related in meaning.

🧩 Main Components

Encoder–Decoder structure (in the original Transformer):

Encoder: Processes input data (e.g., an English sentence).

Decoder: Generates output data (e.g., a French translation).

Each is made of layers with:

Self-Attention → finds relationships between tokens

Feed-Forward Network → applies transformations to each token

Residual connections + Layer Normalization → stabilizes training

Positional Encoding:

Since transformers don’t process tokens in order like RNNs, positional encodings are added to indicate the sequence order.

A key reason for why the architecture is preferred by most modern LLMs is the parallelizability of the architecture over its predecessors. This ensures that the operations necessary for training can be accelerated on a GPU allowing both faster training times and models of bigger sizes to be trained.
