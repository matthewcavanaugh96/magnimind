# Transformers

Transformers are a revolutionary concept in machine learning architecture first posited in the 2017 paper "Attention Is All You Need." They are the backbone of the infrastructure behind ChatGPT.

Transformers first use a method called embedding to convert words into numerical data. Embedding is not new. First, the text is broken down into components called tokens - usually sentences broken into words. 

For each neuron, every token is run through an activation function, which multiplies the value by a common weight to produce a numerical value. This process can be thought of similarly as fitting an intercept to points on a grid.

The embedding process is not exclusive to transformers. It has long been used in traditional Recurrent Neural Networks. But while RNNs work well for sequential or time-based data, they face limitations in text processing. This is because RNNs have to process inputs step by step, so they either break down or become computationally expensive in large datasets.

What makes transformers different is their use of the attention mechanism, which allows the model to learn the relationships between words and how much "attention" each word should pay to each other word. They are trained on a provided corus to see how often each word ppears with another. Recurrence and convolutions are bypassed altogether.

Transformers also use positional encoding to record the order of words, which is massively important in interpreting the meaning of a sentence or paragraph. Positional values are added to the word embeddings, so that a repeated word can be distinguished from other instances of itself.

Each token is assigned a query value which determines what it is searching for in other tokens. From the query value, a key value is calculated for each token (including itself) to calculate its similarity. This process is repeated for each token. The Dot Product represents the similarity between the query value and the key value - the higher the dot product, the more similar, and thus the higher the influence on the word's encoding.

Then, a scaling function is required to convert the Dot Products into values that can represent percentages. The SoftMax function is the most common - it takes all Dot Products and rescales them between 0 and 1 with values that add up to 1. The values are further scaled by their SoftMax (or ArgMax etc) score. These values determine how strongly each token should be influenced by each other token.

Self-attention makes transformers more powerful than RNNs because the meaning of each word has been enriched by context.

These se-flattention values are added to the position-endocded values, to create Residual Values. These are useful because models are not required to remember and process encoding information for every word every time.

The word embedding, positional encoding, self-attention, and residual connection values are constant for each word; the infrastructure of transformers allows much of the calculations to be performed parallel.

In translation tasks, the process continues to an output decoder. A Fully Connected Layer is created in which scaling functions are performed again, and the two scaled values are added to get a Encoder-Decoder Attention Value. This is known as a Fully Connected Layer.

# Old Example

```python
# Install dependencies (uncomment if running in Colab or a new environment)
# !pip install transformers torch

from transformers import pipeline

# Load a pretrained transformer pipeline
classifier = pipeline("sentiment-analysis")

# Provide sample text
texts = [
    "I love using transformers, they make NLP so easy!",
    "This code is frustrating and slow.",
    "The new model is surprisingly accurate.",
    "I thought it was fine, but not great."
]

# Run the model
results = classifier(texts)

# Display results
for text, result in zip(texts, results):
    print(f"Text: {text}")
    print(f"Sentiment: {result['label']}, Confidence: {result['score']:.2f}")
    print("-" * 50)
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

# Load model and tokenizer
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")
model = AutoModelForSequenceClassification.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")

# Tokenize
inputs = tokenizer("Transformers are amazing!", return_tensors="pt")

# Forward pass through the model
outputs = model(**inputs)
logits = outputs.logits
predicted_class_id = logits.argmax().item()

print("Predicted class ID:", predicted_class_id)
print("Label:", model.config.id2label[predicted_class_id])
```
