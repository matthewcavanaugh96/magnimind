```python
features_df = pd.read_csv('features_data.csv')
features_df
```

Below I will add a few columns to provide greater context about each user.

```python
# Date of each user's first trade
features_df['first_trade'] = (
    equity_df
    .groupby('user_id')['date']
    .transform('min')
)

# Date of each user's last trade
features_df['last_trade'] = (
    equity_df
    .groupby('user_id')['date']
    .transform('max')
)

features_df
```

What does risk tolerance look like relative to experience, and vice versa?

```python
risk_experience = features_df.groupby('risk_tolerance')['investment_experience'].value_counts()
risk_experience
```

```python
experience_by_risk = (
    features_df
    .groupby('risk_tolerance')['investment_experience']
    .value_counts(normalize=True)
    .mul(100)
    .rename('percentage')
)

experience_by_risk
```

```python
risk_by_experience = (
    features_df
    .groupby('investment_experience')['risk_tolerance']
    .value_counts(normalize=True)
    .mul(100)
    .rename('percentage')
)

risk_by_experience
```

Low risk tolerance is rare at all levels of experience. The greater trend is that with greater experience, the middle option falls away, and investors are more likely to have higher risk tolerance.
