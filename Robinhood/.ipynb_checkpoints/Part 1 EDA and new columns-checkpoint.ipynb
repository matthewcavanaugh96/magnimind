```python
features_df = pd.read_csv('features_data.csv')
features_df
```

```python
features_df.isnull().sum()
```

Below I will add a few columns to provide greater context about each user.

```python
# Date of each user's first trade
features_df['first_trade'] = (
    equity_df
    .groupby('user_id')['date']
    .transform('min')
)

# Date of each user's last trade
features_df['last_trade'] = (
    equity_df
    .groupby('user_id')['date']
    .transform('max')
)

features_df['first_trade'] = pd.to_datetime(features_df['first_trade'])
features_df['last_trade'] = pd.to_datetime(features_df['last_trade'])


# Total days trading
# Redundant as no user has multiple trades in a single day
# features_df['total_active_days'] = features_df['user_id'].map(
#     equity_df.groupby('user_id')['date'].nunique()
# )

# Total trading span
features_df['trading_span'] = (features_df['last_trade'] - features_df['first_trade']).dt.days

# Total trades
features_df['total_trades'] = (
    features_df['user_id']
    .map(equity_df.groupby('user_id')['timestamp'].count())
)

# Time per trade
features_df['time_per_trade'] = features_df['time_spent'] / features_df['total_trades']

# Trades per day
features_df['trades_per_day'] = features_df['trading_span'] / features_df['total_trades']

# Trades per active day
# Redundant as no user has multiple trades in a single day
# features_df['trades_per_active_day'] = features_df['total_trades'] / features_df['total_active_days']

features_df
```

What does risk tolerance look like relative to experience, and vice versa?

```python
risk_experience = features_df.groupby('risk_tolerance')['investment_experience'].value_counts()
risk_experience
```

```python
experience_by_risk = (
    features_df
    .groupby('risk_tolerance')['investment_experience']
    .value_counts(normalize=True)
    .mul(100)
    .rename('percentage')
)

experience_by_risk
```

```python
risk_by_experience = (
    features_df
    .groupby('investment_experience')['risk_tolerance']
    .value_counts(normalize=True)
    .mul(100)
    .rename('percentage')
)

risk_by_experience
```

Low risk tolerance is rare at all levels of experience. With greater experience, higher risk tolerance instead comes at the risk of medium tolerance.

```python
risk_by_liquidity_needs = (
    features_df
    .groupby('liquidity_needs')['risk_tolerance']
    .value_counts(normalize=True)
    .mul(100)
    .rename('percentage')
)

risk_by_liquidity_needs
```

What about the average number of trades for each level of risk tolerance?

```python
avg_trades_by_risk = (
    features_df
    .groupby('risk_tolerance')['total_trades']
    .mean()
    .reset_index(name='avg_total_trades')
)

avg_trades_by_risk
```

How does time per trade correlate to risk and experience?

```python
time_per_trade_by_risk = (
    features_df
    .groupby('risk_tolerance')['time_per_trade']
    .mean()
    .reset_index(name='avg_time_per_trade')
)

time_per_trade_by_risk
```

Low risk tolerance is correlated with longer times per trade, and trades with high risk tolerance make quicker decisions. However, most of the increase in speed is achieved with the jump from low to medium risk tolerance.

```python
time_per_trade_by_experience = (
    features_df
    .groupby('investment_experience')['time_per_trade']
    .mean()
    .reset_index(name='avg_time_per_trade')
)

time_per_trade_by_experience
```

Traders with extensive experience make quicker decisions, on average, than those with no experience. However, this rise is not perfectly smooth - moderately experienced trades are slower than those with limited experience.
